# 04 - Estimativa de RUL (Vida Util Remanescente)

Regressao linear da tendencia de vibracao, extrapolada ate o limite de fim
de vida da DIN ISO 10816/20816 (7,1 mm/s). Privilegia explicabilidade.
Ver `ml/pipelines/train_rul.py` e ADR 0005.

```python
import sys
sys.path.insert(0, '../pipelines')
from train_rul import train

meta = train()
meta
```

```python
import joblib
import numpy as np
from features import MODELS_DIR, load_dataset

model = joblib.load(MODELS_DIR / meta['artifact'])
df = load_dataset()
seconds = (df['time'] - df['time'].iloc[0]).dt.total_seconds().to_numpy().reshape(-1, 1)
trend = model.predict(seconds)
print('RUL estimado (dias):', meta['rul_days'])
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(df['time'], df['vibration_velocity_rms'], linewidth=0.5, label='vibracao')
ax.plot(df['time'], trend, color='orange', label='tendencia')
ax.axhline(7.1, color='red', linestyle='--', label='limite ISO')
ax.set_title('Tendencia de vibracao e limite de fim de vida')
ax.legend()
plt.show()
```
