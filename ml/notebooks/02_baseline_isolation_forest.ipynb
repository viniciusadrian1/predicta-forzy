# 02 - Baseline de Operacao (Isolation Forest)

Treina o modelo de baseline com dados de operacao normal e inspeciona os
scores de anomalia. Ver `ml/pipelines/train_baseline.py` e ADR 0005.

```python
import sys
sys.path.insert(0, '../pipelines')
from train_baseline import train

meta = train()
meta
```

```python
import joblib
from features import MODELS_DIR, build_features, load_dataset

model = joblib.load(MODELS_DIR / meta['artifact'])
df = load_dataset()
features = build_features(df)
scores = model.score_samples(features.to_numpy())
print('score medio:', scores.mean().round(4))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(df['time'], scores, linewidth=0.6)
ax.set_title('Score do Isolation Forest (menor = mais anomalo)')
plt.show()
```
