Analise exploratoria do dataset historico de 90 dias gerado por

```python
import sys
sys.path.insert(0, '../pipelines')
from features import load_dataset

df = load_dataset()
print('registros:', len(df))
df.head()
```

```python
# Distribuicao das condicoes operacionais
df['label'].value_counts()
```

```python
df[['voltage_v', 'current_a', 'temperature_c', 'vibration_velocity_rms']].describe()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 3))
ax.plot(df['time'], df['vibration_velocity_rms'], linewidth=0.6)
ax.axhline(7.1, color='red', linestyle='--', label='limite ISO 10816')
ax.set_title('Vibracao (velocidade RMS) ao longo de 90 dias')
ax.legend()
plt.show()
```
